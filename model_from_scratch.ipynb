```python
class logisticRegression():
  def __init__(self,linear_rate,no_of_epochs):
    self.linear_rate=linear_rate
    self.no_of_epochs=no_of_epochs

  def fit(self,X,Y):
    self.m,self.n=X.shape
    self.w = np.zeros(self.n)
    self.b=0
    self.X=X
    self.Y=Y

    for i in range(self.no_of_epochs):
      self.update_weight()
  def update_weight(self,):
    y_hat = 1/(1+ np.exp(self.X.dot(self.w) + self.b))
    dw = self.X.T.dot(y_hat - self.Y) / self.m
    db = np.sum(y_hat - self.Y) / self.m

    self.w = self.w - self.linear_rate * dw
    self.b = self.b - self.linear_rate * db



  def predict (self, X_data):
    y_prediction=1/(1+ np.exp(X_data.dot(self.w) + self.b))
    y_pred=np.where(y_prediction>0.5,1,0)

    return  y_pred
```

```python
class lasso_regression():
  def __init__(self,learning_rate,no_of_epochs,regulization):
    self.learning_rate=learning_rate
    self.no_of_epochs=no_of_epochs
    self.regulization=regulization
  def fit(self,x,y):
    self.m,self.n=x.shape
    self.w=np.zeros(self.n)
    self.b=0
    self.x=x
    self.y=y

    for i in range(self.no_of_epochs):
      self.update_weight()
  def update_weight(self,):
    y_pred=self.predict(self.x)
    dw = (-2 / self.m) * self.x.T.dot(self.y - y_pred)
    dw += self.regulization * np.sign(self.w)

    db = (-2 / self.m) * np.sum(self.y - y_pred)

    self.w=self.w-self.learning_rate*dw

    self.b=self.b-self.learning_rate*db

  def predict(self, x_data):
    return self.w.dot(x_data) + self.b
```

```python
import numpy as np

class SVM():
    def __init__(self, learning_rate, no_of_epochs, regulization):
        self.learning_rate = learning_rate
        self.no_of_epochs = no_of_epochs
        self.regulization = regulization

    def fit(self, x, y):
        self.m, self.n = x.shape

        self.w = np.zeros(self.n)
        self.b = 0

        self.x = x
        self.y = y

        for i in range(self.no_of_epochs):
            self.update_weight()

    def update_weight(self):
        y_level = np.where(self.y <= 0, -1, 1)

        for index, x_i in enumerate(self.x):

            condition = y_level[index] * (
                np.dot(x_i, self.w) - self.b
            ) >= 1

            if condition:
                dw = 2 * self.regulization * self.w
                db = 0

            else:
                dw = 2 * self.regulization * self.w - np.dot(
                    x_i, y_level[index]
                )
                db = y_level[index]

            self.w = self.w - self.learning_rate * dw
            self.b = self.b - self.learning_rate * db

    def predict(self, x_data):
        output = np.dot(x_data, self.w) - self.b

        predicted_levels = np.sign(output)

        y_hat = np.where(predicted_levels <= -1, 0, 1)

        return y_hat
```

```python
class KNeighbours:

    def __init__(self, distance_metric):
        self.distance_metric = distance_metric

    def get_distance_metrics(self, training_data_point, test_data_point):

        if self.distance_metric == "euclidean":

            dist = 0

            for i in range(len(training_data_point) - 1):
                dist += (training_data_point[i] - test_data_point[i]) ** 2

            return np.sqrt(dist)

        else:

            dist = 0

            for i in range(len(training_data_point) - 1):
                dist += abs(training_data_point[i] - test_data_point[i])

            return dist

    def nearest_neighbours(self, X_train, test_data, k):

        distance_list = []

        for training_data in X_train:

            distance = self.get_distance_metrics(
                training_data,
                test_data
            )

            distance_list.append((distance, training_data))

        distance_list.sort(key=lambda x: x[0])

        neighbours = []

        for i in range(k):
            neighbours.append(distance_list[i][1])

        return neighbours

    def predict(self, X_train, test_data, k):

        neighbours = self.nearest_neighbours(
            X_train,
            test_data,
            k
        )

        label = []

        for data in neighbours:
            label.append(data[-1])

        predicted_class = statistics.mode(label)

        return predicted_class
```

```python
# k means clustering

import numpy as np
X = np.array([
    [1, 1],
    [2, 1],
    [5, 5],
    [6, 5],
    [9, 1],
    [10, 1]
], dtype=float)
centroids = np.array([
    [1, 1],
    [5, 5],
    [9, 1]
], dtype=float)

K = 3
max_iter = 100

def calculate_distance(point, centroid):
    return np.sqrt(np.sum((point - centroid) ** 2))

def assign_clusters(X, centroids):
    labels = []

    for point in X:
        distances = [
            calculate_distance(point, centroid)
            for centroid in centroids
        ]

        labels.append(np.argmin(distances))

    return np.array(labels)

def update_centroids(X, labels, centroids):
    new_centroids = []

    for k in range(len(centroids)):
        cluster_points = X[labels == k]


        if len(cluster_points) == 0:
            new_centroids.append(centroids[k])
        else:
            new_centroids.append(
                np.mean(cluster_points, axis=0)
            )

    return np.array(new_centroids)

for iteration in range(max_iter):


    labels = assign_clusters(X, centroids)


    new_centroids = update_centroids(
        X, labels, centroids
    )

    print(f"Iteration {iteration + 1}")
    print("Labels:", labels)
    print("Centroids:\n", new_centroids)


    if np.allclose(centroids, new_centroids):
        print("Converged!")
        centroids = new_centroids
        break

    centroids = new_centroids

print("\nFinal Centroids:")
print(centroids)

print("\nFinal Labels:")
print(labels)
```

```python
import numpy as np


class HierarchicalClustering:

    def __init__(self, n_clusters=2):

        self.n_clusters = n_clusters

        self.labels_ = None

        self.clusters_ = None

        self.merge_history_ = []

    def euclidean_distance(self, point1, point2):

        distance = np.sqrt(
            np.sum((point1 - point2) ** 2)
        )

        return distance

    def cluster_distance(self, cluster1, cluster2, X):

        min_distance = float("inf")

        for i in cluster1:

            for j in cluster2:

                distance = self.euclidean_distance(
                    X[i],
                    X[j]
                )

                if distance < min_distance:

                    min_distance = distance

        return min_distance

    def fit(self, X):

        n_samples = len(X)

        # Initially, every point is an individual cluster

        clusters = [
            [i] for i in range(n_samples)
        ]

        self.merge_history_ = []

        while len(clusters) > self.n_clusters:

            min_distance = float("inf")

            pair_to_merge = None

            # Find the closest pair of clusters

            for i in range(len(clusters)):

                for j in range(i + 1, len(clusters)):

                    distance = self.cluster_distance(
                        clusters[i],
                        clusters[j],
                        X
                    )

                    if distance < min_distance:

                        min_distance = distance

                        pair_to_merge = (i, j)

            i, j = pair_to_merge

            # Merge the selected clusters

            merged_cluster = (
                clusters[i] + clusters[j]
            )

            self.merge_history_.append({

                "cluster_1": clusters[i].copy(),

                "cluster_2": clusters[j].copy(),

                "distance": min_distance,

                "merged_cluster": merged_cluster.copy()

            })

            # Remove larger index first

            clusters.pop(j)

            clusters.pop(i)

            # Add merged cluster

            clusters.append(merged_cluster)

        self.clusters_ = clusters

        # Create labels

        self.labels_ = np.zeros(
            n_samples,
            dtype=int
        )

        for label, cluster in enumerate(clusters):

            for index in cluster:

                self.labels_[index] = label

        return self


# Dataset

X = np.array([

    [1, 1],

    [2, 2],

    [8, 8],

    [9, 9]

])


# Create model

model = HierarchicalClustering(
    n_clusters=2
)


# Fit model

model.fit(X)


print("Final Clusters:")

print(model.clusters_)


print("\nLabels:")

print(model.labels_)


print("\nMerge History:")

for merge in model.merge_history_:

    print(merge)
```

```python
import numpy as np


class DBSCAN:

    def __init__(self, eps=1.5, min_samples=2):
        self.eps = eps
        self.min_samples = min_samples
        self.labels_ = None

    def get_neighbors(self, X, i):
        distances = np.sqrt(np.sum((X - X[i]) ** 2, axis=1))
        return list(np.where(distances <= self.eps)[0])

    def expand_cluster(self, X, labels, visited, neighbors, cluster_id):
        i = 0

        while i < len(neighbors):
            p = neighbors[i]

            if not visited[p]:
                visited[p] = True
                new_neighbors = self.get_neighbors(X, p)

                if len(new_neighbors) >= self.min_samples:
                    for n in new_neighbors:
                        if n not in neighbors:
                            neighbors.append(n)

            if labels[p] == -1:
                labels[p] = cluster_id

            i += 1

    def fit(self, X):
        n = len(X)
        visited = np.zeros(n, dtype=bool)
        labels = np.full(n, -1, dtype=int)
        cluster_id = 0

        for i in range(n):
            if visited[i]:
                continue

            visited[i] = True
            neighbors = self.get_neighbors(X, i)

            if len(neighbors) < self.min_samples:
                continue

            labels[i] = cluster_id
            self.expand_cluster(
                X, labels, visited, neighbors, cluster_id
            )
            cluster_id += 1

        self.labels_ = labels
        return self
```
